```python
import pandas as pd
```

```python
df = pd.read_csv("diabetes_dataset.csv")
```

```python
X = df[['Age', 'Pregnancies', 'BMI', 'Glucose', 'BloodPressure', 'HbA1c', 'LDL',
       'HDL', 'Triglycerides', 'WaistCircumference', 'HipCircumference', 'WHR',
       'FamilyHistory', 'DietType', 'Hypertension', 'MedicationUse']]
y = df['Outcome']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 2529)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model_id3 = DecisionTreeClassifier(criterion='entropy')
```

```python
model_id3.fit(X_train, y_train)
```

```python
y_predict = model_id3.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy = accuracy_score(y_test, y_predict)
```
