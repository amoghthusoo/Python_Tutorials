```python
import pandas as pd
```

```python
df = pd.read_csv("https://raw.githubusercontent.com/YBI-Foundation/Dataset/main/Customer%20Purchase.csv")
```

```python
df.head()
```

```python
# df["Education"].unique()
df["Review"].unique()
# df["Purchased"].unique()
```

```python
df["Education"] = df["Education"].replace({"School" : 0, "UG" : 1, "PG" : 2})
```

```python
df["Review"] = df["Review"].replace({"Poor" : 0, "Average" : 1, "Good" : 2})
```

```python
df["Gender"] = df["Gender"].replace({"Male" : 0, "Female" : 1})
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.columns
```

```python
X = df[['Age', 'Gender', 'Education', 'Review']]
y = df['Purchased']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size = 0.8, random_state = 2529)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
model_c45 = DecisionTreeClassifier(criterion='entropy', min_samples_split=5, min_impurity_decrease=0.01)
```

```python
model_c45.fit(X_train, y_train)
```

```python
y_predict = model_c45.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy = accuracy_score(y_test, y_predict)
```
