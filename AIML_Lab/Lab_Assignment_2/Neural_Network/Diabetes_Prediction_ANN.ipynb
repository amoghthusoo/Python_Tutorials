```python
df = pd.read_csv("diabetes_dataset.csv")
```

```python
X = df[['Age', 'Pregnancies', 'BMI', 'Glucose', 'BloodPressure', 'HbA1c', 'LDL',
       'HDL', 'Triglycerides', 'WaistCircumference', 'HipCircumference', 'WHR',
       'FamilyHistory', 'DietType', 'Hypertension', 'MedicationUse']]
y = df['Outcome']
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 2529)
```

```python
from sklearn.neural_network import MLPClassifier
```

```python
model = MLPClassifier(hidden_layer_sizes = (8,), max_iter = 500)
```

```python
model.fit(X_train, y_train)
```

```python
y_predict = model.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
```

```python
accuracy = accuracy_score(y_test, y_predict)
```

```python
accuracy * 100
```

```python
weights = list(model.coefs_)
for weight in weights[0]:
    for w in weight:
        print(w, end=" ")
    print()

print()
for weight in weights[1]:
    for w in weight:
        print(w, end=" ")
    print()
```
